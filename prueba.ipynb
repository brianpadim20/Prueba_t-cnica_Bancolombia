```python
import pandas as pd

# Cargar el archivo Excel, especificando la hoja 'Evolución Contact Center' y que la segunda fila contiene los encabezados
input_file = 'Base de datos Contact Center Dummy.xlsx' 
output_file = 'Nueva_Base_Datos.xlsx'  

# Leer el archivo Excel, usando la segunda fila como encabezado de la hoja especificada
df = pd.read_excel(input_file, sheet_name='Evolucion Contact center', header=1)
```

```python
# Verificar que los encabezados de las columnas estén correctos para la lectura
print("Columnas en el DataFrame:", df.columns.tolist())
```

```python
# Filtrar las columnas deseadas
columnas_deseadas = [
    'Mes', 'Contestadas', 'Abandonadas', 'Entrantes', 'Salida', 
    'Nivel de Servicio', 'Eficiencia', '% Abandono', '% Ocupación', 
    'Promedio Respuesta', 'Prom Conv Entrada', 'Prom Conv Salida', 
    'Prom No listo', 'Promedio Hold', 'Aht Entrada', 'AHT Salida', 'Año'
    ]

# Crear un nuevo DataFrame con las columnas deseadas
nuevo_df = df[columnas_deseadas].copy()
```

```python
# Crea una lista para almacenar los años y un conjunto para los encabezados (que ya fueron leídos y se repiten)
nuevos_años = []
encabezados_leidos = set()

# Itera sobre cada fila en el DataFrame
for index, row in df.iterrows():
    # Verifica si la fila contiene la palabra "Evolución"
    if 'Evolución' in row['Mes']:
        # Extrae el año de la columna "Mes"
        año = row['Mes'].split(' ')[-1]  # Suponiendo que el año es la última palabra
        # Solo agrega el año si no ha sido leído antes
        if año not in encabezados_leidos:
            nuevos_años.append(año)  # Agrega el año a la lista
            encabezados_leidos.add(año)  # Marca el año como leído
        else:
            # Si el año ya ha sido leído, agrega el último año conocido
            nuevos_años.append(nuevos_años[-1] if nuevos_años else None)
    else:
        # Si no es una fila con año, agrega el último año conocido
        nuevos_años.append(nuevos_años[-1] if nuevos_años else None)
        
```

```python
# Agrega la nueva columna "Año" al DataFrame
df['Año'] = nuevos_años
```

```python
# Eliminar filas que contienen 'Total' en la columna 'Mes'
nuevo_df = nuevo_df[~nuevo_df['Mes'].str.contains('Total', na=False)]
```

```python
# Eliminar filas donde aparezca Evolución [año]
años_a_eliminar = ['Evolución 2016', 'Evolución 2017', 'Evolución 2018', 
                'Evolución 2019', 'Evolución 2020']
nuevo_df = nuevo_df[~nuevo_df['Mes'].isin(años_a_eliminar)]
```

```python
# Guardar el nuevo DataFrame en un archivo Excel
nuevo_df.to_excel(output_file, index=False)

print("El archivo Excel ha sido modificado y guardado como:", output_file)
```
